
# 🌲 Section 2: Upload & Index a PDF

## What happens here?

1. Upload your PDF to the PageIndex cloud.
2. PageIndex uses an LLM to read the document structure.
3. Builds a hierarchical **Tree Index** (like a smart Table of Contents).
4. Returns a `doc_id` for all future operations.

## Why NO Chunking?

Instead of cutting the document into arbitrary 500-token pieces, PageIndex respects the document's natural structure.

### Understanding the Concept

**Traditional RAG:**
- PDF is divided into smaller chunks.
- Each chunk contains a fixed number of tokens.
- Sometimes important context gets separated between chunks.

**Vectorless RAG (PageIndex):**
- Does not rely on arbitrary fixed-size chunking.
- Understands the document's natural structure.
- Organizes information into a hierarchical Tree Index.
- Makes it easier to navigate relevant sections.

## Important Terms

- **PDF Upload:** Uploading the document to PageIndex Cloud for processing.
- **LLM:** Reads and understands the document structure.
- **Tree Index:** A hierarchical structure containing document sections and subsections, like a Table of Contents.
- **doc_id:** A unique document identifier returned by PageIndex, used in future operations.

## Overall Flow

PDF Upload → LLM reads document → Tree Index Creation → Returns `doc_id`

**Remember:** PageIndex uses document structure instead of relying on traditional fixed-size text chunks.


````python

# ==========================================
# Upload PDF to PageIndex Cloud
# ==========================================

# PDF file ka path store kar rahe hain
# "./" ka matlab current folder
# PDF aur notebook same folder mein hone chahiye
PDF_PATH = "./pageindex_sample_document.pdf"

# Check karne ke liye ki kaunsi PDF upload hone wali hai
print(f"📤 Uploading: {PDF_PATH}")


# PDF ko PageIndex Cloud par submit/upload kar rahe hain
# pi_client = initialized PageIndex client
# submit_document() = PDF upload karne ka method
# result = API se return hone wala response
result = pi_client.submit_document(PDF_PATH)


# API response se document ki unique ID nikal rahe hain
# result ek dictionary hai aur "doc_id" uski key hai
# Future operations mein isi ID se document identify hoga
doc_id = result["doc_id"]


# Upload successful hone par confirmation display karna
print("✅ Uploaded!")

# Document ki unique ID display karna
print(f"📄 Document ID: {doc_id}")

# ID ko future use ke liye save rakhne ki reminder
print("Save this ID - you'll use it throughout the notebook")

````

Actually flow kya hai?
1. Tum PDF ko submit_document() se PageIndex Cloud par submit karti ho.
2. PageIndex document ko process karke uska Tree Index banata hai.
3. PageIndex response mein doc_id deta hai.
4. Future mein jab tum question poochogi, toh doc_id se PageIndex ko bataogi ki kis document ke Tree Index par kaam karna hai.
5. Retrieval process relevant nodes/sections find karega aur unka content fetch karega.
6. LLM us content ki help se answer generate karega.

````python

# Tree Index ready hone tak status check karna
# PageIndex background mein asynchronously tree build karta hai

print("⏳ Building tree index...")

while True:
    # Document ID se current processing status lena
    status_result = pi_client.get_document(doc_id)

    # Response se status nikalna
    status = status_result.get("status")

    print(f"Status: {status}")

    # Agar processing complete ho gayi
    if status == "completed":
        print("\n✅ Tree index ready!")
        break

    # Agar processing fail ho gayi
    elif status == "failed":
        print("\n❌ Processing failed. Check your PDF.")
        break

    # 5 seconds wait karke dobara status check karna
    time.sleep(5)
````


# 🔎 Section 3: Inspect the Tree Structure

## What the tree looks like:

```text
Document
├── Introduction (pages 1-3)
│   └── Background (pages 1-2)
│
├── Financial Stability (pages 21-31)
│   ├── Monitoring Vulnerabilities (pages 22-28)
│   └── International Cooperation (pages 28-31)
│
└── Conclusion (pages 45-47)
```

## Each node has:

- `node_id` — Unique ID used during retrieval.
- `title` — Section heading.
- `page_index` — Page number in the original PDF.
- `text` — Section summary (when `node_summary=True`).
- `nodes` — Child sections (nested).


````python

# Fetch the complete Tree Index of the document
tree_result = pi_client.get_tree(doc_id, node_summary=True)

# Extract tree data from API response (default empty list if missing)
pageindex_tree = tree_result.get("result", [])

# Count and display top-level sections
print(f"📊 Top-level sections: {len(pageindex_tree)}")

# Display the first node in JSON format (2 spaces indentation)
print("\n🌲 Raw tree (first node):")
print(json.dumps(pageindex_tree[0] if pageindex_tree else {}, indent=2))

# - get_tree() → Document ka complete Tree Index fetch karta hai.
# - doc_id → Batata hai kis document ka tree chahiye.
# - node_summary=True → Nodes ke summaries bhi include karne ko kehta hai.
# - .get("result", []) → API response se tree data nikalta hai; key missing ho toh empty list deta hai.
# - len(pageindex_tree) → Total top-level sections count karta hai.
# - pageindex_tree[0] → First top-level node select karta hai.
# - json.dumps(..., indent=2) → JSON ko readable format mein print karta hai.
````

````python
# Full Tree Structure ko readable format mein print karna

def print_tree(nodes, indent=0):
    # Har node ko ek-ek karke access karna
    for node in nodes:

        # Tree mein child nodes ko indentation dena
        prefix = "  " * indent + ("└── " if indent > 0 else "")

        # Node ka page number lena
        page = f"{node.get('start_index', '?')}-{node.get('end_index', '?')}"

        # Node ki ID, title aur page number print karna
        print(f"{prefix}[{node['node_id']}] {node['title']} (p.{page})")

        # Agar node ke andar child nodes hain
        if node.get("nodes"):

            # Function ko dobara call karke child nodes print karna
            print_tree(node["nodes"], indent + 1)


# Complete document ka tree display karna
print("📚 Full Document Structure:\n")
print_tree(pageindex_tree)
````

````python
# Tree ke total nodes count karne ka function
def count_nodes(nodes):

    total = len(nodes)  # Current level ke nodes count

    for n in nodes:  # Har node ko access karna

        # Agar node ke andar child nodes hain
        if n.get("nodes"):

            # Child nodes ko recursively count karna
            total += count_nodes(n["nodes"])

    return total  # Total count return karna


# Complete tree ke nodes count karna
total = count_nodes(pageindex_tree)

# Total nodes display karna
print(f"🔢 Total nodes in tree: {total}")
print("Each node = one retrievable section of the document")
````


# How PageIndex Differs from Vector RAG

This is where PageIndex fundamentally differs from Vector RAG.

## 1. Vector RAG Retrieval

```text
Query → Embed → cosine_similarity(query_vec, all_chunk_vecs) → Top-K Chunks
```

**Problem:** Finds what's similar, not necessarily what's relevant.

## 2. PageIndex Retrieval

```text
Query + Tree → LLM Reasons → "Node 0007 and 0008 contain the answer"
```

**Advantage:** LLM understands document structure, context, and intent.

**The LLM acts like a human expert scanning a Table of Contents.**


````python
import json

# Query ke according relevant document nodes find karna
def llm_tree_search(query: str, tree: list, model: str = "openai/gpt-oss-120b") -> dict:

    # Tree ko compress karna, taaki kam tokens use hon
    def compress(nodes):
        out = []

        for n in nodes:
            # Har node ki important information store karna
            entry = {
                "node_id": n["node_id"],
                "title": n["title"],
                "page": n.get("page_index", n.get("start_index", "?")),
                "summary": n.get("text", "")[:150]  # First 150 characters
            }

            # Agar child nodes hain toh unhe bhi compress karna
            if n.get("nodes"):
                entry["children"] = compress(n["nodes"])

            out.append(entry)

        return out  # Compressed tree return karna

    # Original tree ka compressed version banana
    compressed_tree = compress(tree)

    # LLM ke liye prompt banana
    prompt = f"""
You are given a query and a document's tree structure (like a Table of Contents).

Your task: Identify which node IDs most likely contain the answer.
Think step-by-step about which sections are relevant.

Query: {query}

Document Tree:
{json.dumps(compressed_tree, indent=2)}

Reply ONLY in this exact JSON format:
{{
    "thinking": "Brief explanation of relevant sections",
    "node_list": ["node_id1", "node_id2"]
}}
"""

    # Groq LLM ko prompt bhejna
    response = groq_client.chat.completions.create(
        model=model,
        messages=[
            {"role": "user", "content": prompt}
        ],
        response_format={"type": "json_object"}
    )

    # LLM ke JSON response ko Python dictionary mein convert karna
    return json.loads(response.choices[0].message.content)
````

````python
# Test with a query related to our PDF
query = "What is Vectorless RAG and how does tree-based indexing help in document retrieval?"

# Display user query
print(f"🔍 Query: {query}\n")

# Find relevant nodes using LLM Tree Search
result = llm_tree_search(query, pageindex_tree)

# Display LLM reasoning
print("🧠 LLM Reasoning:")
print(result.get("thinking", "N/A"))

# Display selected node IDs
print("\n🎯 Selected Node IDs:", result.get("node_list", []))
````

🚀 End-to-End Vectorless RAG Pipeline (PageIndex)
1. PDF Upload & Tree Indexing
- Upload PDF to PageIndex Cloud using submit_document().
- PageIndex reads the document structure.
- LLM creates a hierarchical Tree Index.
- Each section becomes a node containing:
  - node_id → Unique node ID.
  - title → Section heading.
  - page_index → Original PDF page number.
  - text → Section summary/content.
  - nodes → Child sections.
- Returns doc_id to identify the uploaded document.
2. Wait for Processing
- Use get_document(doc_id) to check document status.
- Keep checking until status becomes completed.
- Tree Index is now ready.
3. Fetch Tree Structure
- Use get_tree(doc_id, node_summary=True).
- Fetch the complete hierarchical Tree Index.
- Store the tree in pageindex_tree.
- Inspect sections, nodes and page numbers.
4. Tree Compression
- Use compress() function to reduce the tree size.
- Keep only important information:
  - node_id
  - title
  - page
  - summary
- Recursively compress child nodes.
- This saves tokens when sending the tree to the LLM.
5. User Query
- User asks a question about the PDF.
- Example: "What is Vectorless RAG?"
6. Tree Search (Retrieval)
- Send user query and compressed document tree to the LLM.
- LLM understands the document structure and identifies relevant sections.
- Selects relevant node_id values.
- Returns selected node IDs and reasoning.
7. Groq LLM API Call
- Use groq_client.chat.completions.create() to send the prompt to Groq.
- Groq returns a JSON response.
- Use json.loads() to convert the JSON response into a Python dictionary.
- Display selected node IDs and LLM reasoning using print().

````python
# Step 8: LLM se mile node IDs ko original tree me find karna
def find_nodes_by_ids(nodes, target_ids):
    found = []

    for node in nodes:
        # Agar node ID match kare, to us node ko store karo
        if node["node_id"] in target_ids:
            found.append(node)

        # Child nodes me bhi search karo
        if node.get("nodes"):
            found.extend(find_nodes_by_ids(node["nodes"], target_ids))

    return found

# Step 7 ke LLM search result se selected node IDs lena
target_ids = result["node_list"]

# Un IDs ke corresponding nodes original tree se find karna
selected_nodes = find_nodes_by_ids(pageindex_tree, target_ids)
````

````python
# Connection: Step 8 me relevant nodes mile. 
# Step 9 un nodes ka content context me daalta hai aur Groq LLM se final answer generate karwata hai.

# Step 9: Selected nodes se final answer generate karna
def generate_answer(query, nodes, model="openai/gpt-oss-120b"):

    # Selected nodes ka content context me collect karna
    context = ""

    for node in nodes:
        context += f"""
        Title: {node.get('title', '')}
        Content: {node.get('text', node.get('summary', ''))}
        """

    # Query aur context ke basis par answer generate karna
    prompt = f"""
    Answer the question using the given context.

    Context:
    {context}

    Question: {query}
    """

    response = groq_client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}]
    )

    return response.choices[0].message.content
````

````python
# Connection: Ye function Step 7, 8 aur 9 ko sequence me run karta hai. 
# Hume bas query deni hai; baaki process automatically execute hoga.
# Step 10: Saare steps ko ek function me connect karna
def vectorless_rag(query, tree, verbose=True):

    # Step 7: Query ke according relevant node IDs find karna
    result = llm_tree_search(query, tree)

    # Step 8: Selected IDs ke corresponding nodes find karna
    selected_nodes = find_nodes_by_ids(tree, result["node_list"])

    # Step 9: Selected nodes se final answer generate karna
    answer = generate_answer(query, selected_nodes)

    if verbose:
        print("Selected Node IDs:", result["node_list"])
        print("Final Answer:", answer)

    return answer
````

Traditional RAG vs Vectorless RAG – Hinglish Notes
Pehle dono ka basic difference samjho:

Traditional RAG: Document ko chhote-chhote chunks mein divide karta hai, unki embeddings banata hai aur Vector Database mein store karta hai. Query aane par similar chunks search karta hai.

Vectorless RAG: Embeddings aur Vector Database use nahi karta. Document ka Tree Structure banata hai aur LLM us structure ko navigate karke relevant information find karta hai.

1. Traditional RAG – Kab Use Karein?
Best for: Fast retrieval + Large data
1. Massive Heterogeneous Corpora

Jab bahut saare documents different formats mein hon.

Example: Millions of blogs, PDFs, tickets, transcripts aur knowledge-base articles.

Traditional RAG large aur mixed data handle karne ke liye useful hai.

2. Latency-Critical Applications

Jab application ko bahut fast response dena ho.

Example: Chatbots, voice assistants aur search-as-you-type.

Isme har millisecond important hota hai.

3. Short Factoid Queries

Jab user short aur direct questions poochta hai.

Example: "Who is the CEO?" ya "Warranty period kitna hai?"

Answer usually ek hi chunk mein mil jata hai.

4. Cost-Sensitive at Scale

Jab daily thousands of queries aati hain aur retrieval cost kam rakhni ho.

Traditional RAG mein vector similarity search fast aur relatively cheap hoti hai.

2. Vectorless RAG – Kab Use Karein?
Best for: Structured documents + Deep reasoning
1. Long, Structured Documents

Jab documents bade hon aur properly headings, chapters aur sections mein divided hon.

Example: Annual reports, legal contracts, research papers, regulatory filings aur textbooks.

Tree Structure ki help se relevant section tak pahunch sakte hain.

2. Reasoning > Similarity

Jab sirf similar words search nahi karne, balki different sections ko compare aur understand karna ho.

Example: "Section 7 ke risk factors ko Section 12 ke mitigations se compare karo."

LLM dono sections ko navigate karke reasoning kar sakta hai.

3. Explainability is Required

Jab answer ke saath ye bhi batana ho ki information kis section se li gayi hai.

Example: Legal, financial advisory, compliance aur audit.

Navigation path se explain kar sakte hain ki answer kahan se aaya.

4. Chunking Destroys Context

Jab document ko chhote chunks mein divide karne se important information ka connection break ho jata hai.

Example: Ek section mein kisi term ki definition hai aur doosre section mein uska reference.

Vectorless RAG full section read karke context preserve kar sakta hai.

3. Final Comparison – Quick Revision
Traditional RAG

Vectorless RAG

Chunk-based retrieval

Tree-based navigation

Embeddings + Vector DB

No embeddings / Vector DB

Large, mixed documents

Long, structured documents

Fast retrieval

Deeper reasoning

Short factual questions

Cross-section questions

Similarity search

LLM reasoning

Low retrieval cost

Higher LLM cost

Millions of documents

Tens of thousands (as shown in slides)

4. Easy Example – Dono Mein Difference
Suppose tumhare paas ek 100-page annual report hai.

Question 1: "Company ka revenue kitna hai?"

Traditional RAG: Revenue wale chunk ko similarity search se find karega.

Vectorless RAG: Tree mein Financials section navigate karke answer find karega.

Question 2: "Company ke risk factors aur unke solutions compare karo."

Traditional RAG: Dono sections ke chunks retrieve kar sakta hai, lekin unka connection samajhna difficult ho sakta hai.

Vectorless RAG: Risk aur Mitigation sections navigate karke dono ko compare kar sakta hai.

Yaad rakhna:

Traditional RAG = Search by Similarity

Vectorless RAG = Navigate by Structure
