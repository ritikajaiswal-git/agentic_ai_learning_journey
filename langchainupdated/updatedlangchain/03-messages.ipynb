# message prompt 

## Message Prompts

Message prompts mein hum model ko ek single text string dene ke bajaye, messages ki list dete hain. Har message ka apna role aur content hota hai.

### Message Types

- **System Message:** Model ko instructions deta hai ki kaise behave karna hai, kis role mein answer dena hai aur kya guidelines follow karni hain.
- **Human Message:** User ka input ya question represent karta hai. Ismein text, images, audio ya files ho sakti hain.
- **AI Message:** Model ka response represent karta hai. Ismein text, tool calls aur extra metadata ho sakta hai.
- **Tool Message:** Tool execute hone ke baad jo result milta hai, use model tak wapas pahunchata hai.

## System Message

System message model ke behaviour ko control karne ke liye initial instructions deta hai. Isse model ka role, response ka tone aur answer dene ki guidelines set kar sakte hain.

**Example:** "You are a helpful Python teacher. Explain everything in simple language."

## Human Message

Human message user ke question ya input ko represent karta hai. User text, images, audio aur doosra supported content bhej sakta hai.

**Example:** "What is Python?"

## AI Message

AI message model ke response ko represent karta hai. Jab hum model ko invoke karte hain, toh uska response AI message ke form mein mil sakta hai.

**Example:** "Python is a programming language."

AI message mein tool call ki request aur provider-specific metadata bhi ho sakta hai.

## Tool Message

Tool message kisi tool ke execute hone ke baad uska result contain karta hai. Ye result model ko diya jata hai, taaki model uske basis par final answer bana sake.

**Example:**
- AI Message: "get_weather tool ko Boston ke liye call karo."
- Tool Message: "72°F, sunny."
- AI Message: "Boston mein weather 72°F aur sunny hai."


```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

# Messages ki list banana: system instructions aur user ka question
messages = [
    # SystemMessage model ko role aur instructions deta hai
    SystemMessage("You are a poetry expert"),

    # HumanMessage user ka actual question ya request deta hai
    HumanMessage("Write a poem on artificial intelligence")
]

# Messages model ko bhejna aur response receive karna
response = model.invoke(messages)

# AI ke response mein se sirf poem ka text nikalna
response.content
```

```python
# SystemMessage aur HumanMessage pehle import hone chahiye

# Step 1: Model ko coding assistant ka role dena
system_msg = SystemMessage("You are a helpful coding assistant.")

# Step 2: System instruction aur user ka question messages list mein rakhna
messages = [
    system_msg,
    HumanMessage("How do I create a REST API?")
]

# Step 3: Messages model ko bhejna aur response receive karna
response = model.invoke(messages)

# Step 4: Model ke answer ka text print karna
print(response.content)
```

```python
# SystemMessage aur HumanMessage import karna
from langchain.messages import SystemMessage, HumanMessage

# Step 1: Model ko detailed instructions dena
system_msg = SystemMessage("""
You are a senior Python developer with expertise in web frameworks.
Always provide code examples and explain your reasoning.
Be concise but thorough in your explanations.
""")

# Step 2: System instructions aur user ka question messages list mein rakhna
messages = [
    system_msg,
    HumanMessage("How do I create a REST API?")
]

# Step 3: Messages model ko bhejna aur response receive karna
response = model.invoke(messages)

# Step 4: Model ke answer ka text print karna
print(response.content)
```

```python
# HumanMessage create karna aur usmein metadata add karna
human_msg = HumanMessage(
    content="Hello!",  # User ka actual message
    name="alice",      # User ko identify karne ke liye optional name
    id="msg_123"       # Message ko uniquely identify karne ke liye optional ID
)

# HumanMessage ko model ko bhejna
response = model.invoke([
    human_msg
])

# Complete AIMessage response display karna
response
```

```python
from langchain.messages import AIMessage, SystemMessage, HumanMessage

# AI ka response manually create karna
# Maan rahe hain ki AI ne pehle ye reply diya tha
ai_msg = AIMessage("I'd be happy to help you with that question!")

# Conversation history banana
messages = [
    # Model ko batana ki use helpful assistant ki tarah behave karna hai
    SystemMessage("You are a helpful assistant"),

    # User ka pehla question
    HumanMessage("Can you help me?"),

    # Pehle ke AI response ko conversation history mein add karna
    ai_msg,

    # User ka agla question
    HumanMessage("Great! What's 2+2?")
]

# Puri conversation history model ko dena
response = model.invoke(messages)

# Model ke final answer ka text print karna
print(response.content)
```

```python
#The weather in San Francisco is sunny, with a temperature of 72°F.
from langchain.messages import AIMessage, ToolMessage, HumanMessage

# Step 1: AI ka tool-call response manually create karna
ai_message = AIMessage(
    content=[],  # AI ne normal text answer nahi diya
    tool_calls=[{
        "name": "get_weather",  # AI jis tool ko call karna chahta hai
        "args": {"location": "San Francisco"},  # Tool ko diye gaye arguments
        "id": "call_123"  # Tool call ki unique ID
    }]
)

# Step 2: Tool execution ka result maan kar store karna
# Yahan actual API call nahi ho rahi; result manually diya gaya hai
weather_result = "Sunny, 72°F"

# Tool ke result ka message banana
tool_message = ToolMessage(
    content=weather_result,  # Tool se mila result
    tool_call_id="call_123"  # Ye ID AI ke tool-call ID se match honi chahiye
)

# Step 3: Conversation history ko continue karna
messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_message,     # AI ki tool-call request
    tool_message    # Tool ka result
]

# Step 4: Conversation model ko dobara dena
# Model tool result ko samajhkar final answer dega
response = model.invoke(messages)

# Final answer ka text display karna
print(response.content)
```

Is code se tumhe LangChain mein Tool Calling ka message flow samajhne ko milta hai.
3 main cheezein seekhne ko milti hain
1. AIMessage — AI batata hai ki kaunsa tool call karna hai aur kaunse arguments dene hain.

2. ToolMessage — Tool execute hone ke baad uska result store karta hai. tool_call_id se result ko uski original tool request se match kiya jata hai.

3. model.invoke(messages) — AI ki request aur tool ka result model ko dobara deta hai, taaki woh final answer bana sake.
