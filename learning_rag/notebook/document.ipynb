```python
from langchain_core.documents import Document

# Document object create kar rhe hain
doc = Document(
    page_content="this is the main text content I am using to create RAG",

    # Metadata stores additional information about the document
    metadata={
        "source": "example.txt",  # Document kis file se aaya hai
        "pages": 1,                # Document ka page number
        "author": "Krish Naik",    # Document ka author
        "date_created": "2025-01-01"  # Document kab create hua
    }
)

# WHY METADATA IS IMPORTANT?
# Suppose user uploads 2-3 different documents.
# User asks: "What is RAG written by Krish Naik?"
# Metadata filter author="Krish Naik" lagayega.
# Isse sirf Krish Naik ke document ka relevant data retrieve hoga.
# Baaki authors ke documents ignore ho jayenge.
# Thus, metadata helps in filtering and retrieving the correct document.    
```

```python
import os

# .  -> Current directory (notebook folder)
# .. -> Parent directory (learning_RAG folder)
# ../data/text_files -> Go one folder back, then enter data/text_files
# exist_ok=True -> No error if folder already exists

os.makedirs("../data/text_files", exist_ok=True)
```

```python



# Step 2: Store sample text data in a dictionary
# Key = File path
# Value = File ke andar ka actual content

sample_texts = {

    "../data/text_files/python_intro.txt": """Python Programming Introduction

Python is a high-level, interpreted programming language known for its simplicity and readability.
Created by Guido van Rossum and first released in 1991, Python has become one of the most popular
programming languages in the world.

Key Features:
- Easy to learn and use
- Extensive standard library
- Cross-platform compatibility
- Strong community support

Python is widely used in web development, data science, artificial intelligence, and automation.
""",
# us data k andr text file k andr 2 file bna , ek machine learning ar ek python intro
    "../data/text_files/machine_learning.txt": """Machine Learning Introduction

Machine Learning is a branch of Artificial Intelligence (AI) that allows computers to learn from data
and make predictions or decisions without being explicitly programmed.

Types of Machine Learning:

1. Supervised Learning: Learning with labeled data.
2. Unsupervised Learning: Finding patterns in unlabeled data.
3. Reinforcement Learning: Learning through rewards and penalties.

Applications include image recognition, speech processing, and recommendation systems.
"""
}


# Step 3: Iterate over dictionary items
# filepath = File ka path
# content = File ke andar ka data

for filepath, content in sample_texts.items():

    # Step 4: Open each file in write mode
    # 'w' -> Write mode (create or overwrite file)
    # encoding="utf-8" -> Supports different characters

    with open(filepath, 'w', encoding="utf-8") as f:

        # Step 5: Write content into the file
        f.write(content)


# Step 6: Print confirmation message
print("✅ Sample text files created!")
```

```python
# DirectoryLoader: Folder ke andar multiple files load karne ke liye

from langchain_community.document_loaders import DirectoryLoader
from langchain_community.document_loaders import TextLoader

#from langchain_community.document_loaders import PyPDFLoader(iska kam avi ni h , pdf ncihe m kam ayga)


# Step 1: Create DirectoryLoader object
dir_loader = DirectoryLoader(
    "../data/text_files",  # Folder ka path

    glob="**/*.txt",  # Folder aur subfolders ki sabhi .txt files match karega, jo txt file h usko hi dega folder s

    loader_cls=TextLoader,  # Har matched file ko read karne ke liye TextLoader use hoga

    loader_kwargs={'encoding': 'utf-8'},  # UTF-8 encoding (readable char txt form m , byte k form m ni)

    show_progress=False  # Progress bar show nahi karna
)

# Step 2: Load all matched files
documents = dir_loader.load()

# Step 3: Display loaded documents
documents


# IMPORTANT:
# TextLoader -> Single .txt file load karta hai.
# DirectoryLoader -> Ek folder se multiple files load karta hai.
#
# PDF ke case mein:
# PyPDFLoader -> Single PDF file load karta hai.
# DirectoryLoader + PyPDFLoader -> Folder se multiple PDF files load karta hai.
#
# PDF files ke liye:
# glob="**/*.pdf"
# loader_cls=PyPDFLoader
#
# Note: loader_cls mein TextLoader ki jagah PyPDFLoader use karenge.
```

```python
from langchain_community.document_loaders import DirectoryLoader, PyMuPDFLoader

# DirectoryLoader -> Folder ki multiple PDF files load karega
dir_loader = DirectoryLoader(
    "../data/pdf",  # PDF files wale folder ka path

    glob="**/*.pdf",  # All PDF files ko match karega

    loader_cls=PyMuPDFLoader,  # PDF read karne ke liye PyMuPDFLoader

    # encoding ki zarurat nahi hai PyMuPDFLoader mein

    show_progress=False  # Progress bar hide rakhega
)

# Load all PDF documents
documents = dir_loader.load()

# Display loaded documents
documents
```
